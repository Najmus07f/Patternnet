```python

import os, warnings, math
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import RobustScaler
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve, auc
from sklearn.utils.class_weight import compute_class_weight
from sklearn.preprocessing import label_binarize

import tensorflow as tf
from tensorflow.keras import layers, Model

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

CLASS_NAMES = ["Noise", "Consolidation", "DoubleTop"]
NUM_CLASSES = 3

# ---------------------------
# 0) LOAD DSE CSV
# ---------------------------
csv_path = "/kaggle/input/dhaka-stock-exchange-dataset/DSE_Data.csv"
if not os.path.exists(csv_path):
    raise FileNotFoundError("DSE_Data.csv not found at /kaggle/input/dhaka-stock-exchange-dataset/DSE_Data.csv")

df = pd.read_csv(csv_path)
print("Loaded:", csv_path, "| rows:", len(df), "| cols:", list(df.columns))

need = {"Trading_Code","Date","Open","High","Low","Close","Volume"}
if not need.issubset(df.columns):
    raise ValueError(f"Dataset must have {need}. Missing: {need - set(df.columns)}")

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df = df.dropna(subset=["Date","Trading_Code","Open","High","Low","Close","Volume"]).copy()

for c in ["Open","High","Low","Close","Volume"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")
df = df.dropna(subset=["Open","High","Low","Close","Volume"]).copy()

df = df[(df["High"] >= df["Low"]) & (df["Volume"] >= 0)].copy()
df = df.sort_values(["Trading_Code","Date"]).reset_index(drop=True)

print("Date range:", df["Date"].min(), "->", df["Date"].max())
print("Tickers:", df["Trading_Code"].nunique())

# ---------------------------
# 1) Leak-safe per-row features (per ticker)
# ---------------------------
g = df.groupby("Trading_Code", group_keys=False)

# log return (close-to-close)
df["logret"] = g["Close"].apply(lambda s: np.log(np.maximum(s, 1e-9)).diff()).fillna(0.0)

# candle geometry raw
df["range_hl"]  = (df["High"] - df["Low"]).astype(float)
df["body"]      = (df["Close"] - df["Open"]).astype(float)
df["upper_wick"]= (df["High"] - np.maximum(df["Open"], df["Close"])).astype(float)
df["lower_wick"]= (np.minimum(df["Open"], df["Close"]) - df["Low"]).astype(float)

def true_range(sub):
    prev_close = sub["Close"].shift(1)
    tr = np.maximum(sub["High"] - sub["Low"],
         np.maximum(np.abs(sub["High"] - prev_close), np.abs(sub["Low"] - prev_close)))
    return tr

df["tr"] = g.apply(true_range).reset_index(level=0, drop=True).fillna(df["range_hl"])

# SHIFT(1) to avoid peeking current bar stats into itself
df["atr14"] = g["tr"].apply(lambda s: s.rolling(14, min_periods=5).mean().shift(1)).fillna(method="bfill").fillna(0.0)
df["rv14"]  = g["logret"].apply(lambda s: s.rolling(14, min_periods=5).std().shift(1)).fillna(method="bfill").fillna(0.0)

df["logvol"]   = np.log1p(np.maximum(df["Volume"].astype(float), 0.0))
df["relvol20"] = g["logvol"].apply(lambda s: (s - s.rolling(20, min_periods=8).median()).shift(1)).fillna(0.0)

# scale-invariant ratios
df["atr_ratio"]   = (df["atr14"] / np.maximum(df["Close"], 1e-9)).astype(float)
df["range_ratio"] = (df["range_hl"] / np.maximum(df["Close"], 1e-9)).astype(float)

# ---------------------------
# 2) Time encodings (daily market)
# ---------------------------
dow   = df["Date"].dt.dayofweek.astype(int)
month = df["Date"].dt.month.astype(int)

df["sin_week"]  = np.sin(2*np.pi * dow / 7.0)
df["cos_week"]  = np.cos(2*np.pi * dow / 7.0)
df["sin_month"] = np.sin(2*np.pi * month / 12.0)
df["cos_month"] = np.cos(2*np.pi * month / 12.0)

# ---------------------------
# 3) Helpers (labels + engineered features)  [FIXED DT trough]
# ---------------------------
def slope(y):
    x = np.arange(len(y))
    return np.polyfit(x, y, 1)[0]

def fit_rmse(y):
    x = np.arange(len(y))
    m, b = np.polyfit(x, y, 1)
    yhat = m*x + b
    return float(np.sqrt(np.mean((y-yhat)**2)))

def range_contraction(highs, lows):
    r = highs - lows
    return -slope(r)

def local_peaks(y, min_sep=4):
    idx=[]
    for i in range(1,len(y)-1):
        if y[i] > y[i-1] and y[i] > y[i+1]:
            idx.append(i)
    filt=[]
    for i in idx:
        if not filt or (i - filt[-1] >= min_sep):
            filt.append(i)
    return filt

def normalize_window(highs, lows, opens=None, closes=None):
    mid = (highs + lows)/2.0
    base = np.median(mid)
    scale = (np.percentile(highs, 95) - np.percentile(lows, 5))
    scale = scale if scale > 1e-9 else 1.0
    out = {
        "H": ((highs - base)/scale).astype(np.float32),
        "L": ((lows  - base)/scale).astype(np.float32),
        "base": float(base),
        "scale": float(scale),
    }
    if opens is not None:
        out["O"] = ((opens - base)/scale).astype(np.float32)
    if closes is not None:
        out["C"] = ((closes - base)/scale).astype(np.float32)
    return out

# SAME 3-class labels: 0 Noise, 1 Consolidation, 2 DoubleTop
def label_window_3class(highs, lows, level,
                        eps_slope=0.0005, rmse_frac=0.25, dip=0.004, min_conf=1.9):
    """
    FIXED DoubleTop:
    - trough from LOW valley between peaks
    - dip threshold is level-based: dip * level
    """
    total_range = float(np.max(highs) - np.min(lows))
    if total_range < level*0.002:
        return 0

    up_s  = slope(highs)
    low_s = slope(lows)
    contr = range_contraction(highs, lows)

    rmse_h = fit_rmse(highs)
    rmse_l = fit_rmse(lows)
    rmse_gate = total_range * rmse_frac
    fit_ok = (rmse_h <= rmse_gate) and (rmse_l <= rmse_gate)

    tol = max(level*0.0015, 1e-9)

    # Consolidation score
    conv = 0.0
    conv += 1.2 if contr > 0 else -1.2
    if up_s < -eps_slope and low_s > eps_slope:
        conv += 1.3
    if abs(up_s - low_s) > max(abs(up_s), abs(low_s))*0.25:
        conv += 0.5
    conv += 0.6 if fit_ok else -0.8

    # DoubleTop score
    dt = 0.0
    pk = local_peaks(highs, min_sep=4)
    if len(pk) >= 2:
        best_pair, best_diff = None, 1e18
        for a in range(len(pk)):
            for b in range(a+1, len(pk)):
                i1, i2 = pk[a], pk[b]
                if i2 - i1 < 6:
                    continue
                diff = abs(highs[i1] - highs[i2])
                if diff < best_diff:
                    best_diff = diff
                    best_pair = (i1, i2)

        if best_pair is not None:
            i1, i2 = best_pair
            p1, p2 = highs[i1], highs[i2]

            # ✅ trough from LOW valley
            trough = float(np.min(lows[i1:i2+1]))

            # peak similarity
            if abs(p1 - p2) <= tol:
                dt += 1.4

            # ✅ level-based dip threshold
            if (min(p1, p2) - trough) >= (dip * level):
                dt += 1.4

            # rebound confirmation (optional)
            if highs[-1] > trough + tol:
                dt += 0.3

    scores = {1: conv, 2: dt}
    best_label = max(scores, key=scores.get)
    best_score = float(scores[best_label])
    return best_label if best_score >= min_conf else 0

def engineered_features(highs, lows, level, atr_w, rv_w, relvol_w):
    """
    FIXED DoubleTop geo:
    - trough_depth from LOW valley between peaks
    """
    up_s  = slope(highs)
    low_s = slope(lows)
    contr = range_contraction(highs, lows)

    rmse_h = fit_rmse(highs)/max(level,1e-9)
    rmse_l = fit_rmse(lows) /max(level,1e-9)
    r = highs - lows

    pk = local_peaks(highs, min_sep=4)
    peak_diff = 1.0
    peak_gap = 0.0
    trough_depth = 0.0
    if len(pk) >= 2:
        best_pair, best_diff = None, 1e18
        for a in range(len(pk)):
            for b in range(a+1, len(pk)):
                i1, i2 = pk[a], pk[b]
                if i2 - i1 < 6:
                    continue
                diff = abs(highs[i1] - highs[i2])
                if diff < best_diff:
                    best_diff = diff
                    best_pair = (i1,i2)

        if best_pair is not None:
            i1,i2 = best_pair
            p1,p2 = highs[i1], highs[i2]

            # ✅ trough from LOW valley
            trough = float(np.min(lows[i1:i2+1]))

            peak_diff = float(abs(p1-p2)/max(level,1e-9))
            peak_gap  = float((i2-i1)/len(highs))
            trough_depth = float((min(p1,p2)-trough)/max(level,1e-9))

    # DSE summary (Noise vs Consol separation support)
    atr_mean     = float(np.mean(atr_w))
    atr_trend    = float(slope(atr_w)) if len(atr_w) > 3 else 0.0
    rv_mean      = float(np.mean(rv_w))
    relvol_mean  = float(np.mean(relvol_w))
    relvol_trend = float(slope(relvol_w)) if len(relvol_w) > 3 else 0.0

    return np.array([
        up_s, low_s, contr,
        rmse_h, rmse_l,
        float(np.std(r)/max(level,1e-9)),
        peak_diff, peak_gap, trough_depth,
        atr_mean, atr_trend, rv_mean, relvol_mean, relvol_trend
    ], dtype=np.float32)

# ---------------------------
# 4) Build per-ticker windows (sequence + geo + label + start time)
# ---------------------------
WINDOW = 30
STRIDE = 30
MIN_CONF = 1.9

X_seq, X_geo, Y, TSTART = [], [], [], []

for code, sub in df.groupby("Trading_Code"):
    sub = sub.sort_values("Date").reset_index(drop=True)
    if len(sub) < WINDOW + 5:
        continue

    for start in range(0, len(sub) - WINDOW, STRIDE):
        w = sub.iloc[start:start+WINDOW]

        highs  = w["High"].to_numpy(float)
        lows   = w["Low"].to_numpy(float)
        opens  = w["Open"].to_numpy(float)
        closes = w["Close"].to_numpy(float)

        level = float(np.median((highs + lows)/2.0))
        if not np.isfinite(level) or level <= 0:
            continue

        y = label_window_3class(highs, lows, level, min_conf=MIN_CONF)

        normed = normalize_window(highs, lows, opens=opens, closes=closes)
        H, L = normed["H"], normed["L"]
        O, C = normed["O"], normed["C"]

        body_n  = (C - O).astype(np.float32)
        range_n = (H - L).astype(np.float32)
        up_w    = (H - np.maximum(O, C)).astype(np.float32)
        lo_w    = (np.minimum(O, C) - L).astype(np.float32)

        mid = (highs + lows)/2.0
        ret = np.diff(mid) / np.maximum(mid[:-1], 1e-9)
        ret = np.concatenate([[0.0], ret]).astype(np.float32)
        rscale = np.std(ret) if np.std(ret) > 1e-9 else 1.0
        ret_n = (ret / rscale).astype(np.float32)

        atr_w  = w["atr_ratio"].to_numpy(np.float32)
        rv_w   = w["rv14"].to_numpy(np.float32)
        rv_w   = (rv_w / (np.std(rv_w)+1e-9)).astype(np.float32)
        relv_w = w["relvol20"].to_numpy(np.float32)

        sinw = w["sin_week"].to_numpy(np.float32)
        cosw = w["cos_week"].to_numpy(np.float32)
        sinm = w["sin_month"].to_numpy(np.float32)
        cosm = w["cos_month"].to_numpy(np.float32)

        chans = [
            H, L, O, C,
            range_n, body_n, up_w, lo_w,
            ret_n,
            atr_w, rv_w, relv_w,
            sinw, cosw, sinm, cosm
        ]
        seq = np.stack(chans, axis=-1).astype(np.float32)

        geo = engineered_features(
            highs, lows, level,
            atr_w=w["atr_ratio"].to_numpy(float),
            rv_w=w["rv14"].to_numpy(float),
            relvol_w=w["relvol20"].to_numpy(float)
        )

        X_seq.append(seq)
        X_geo.append(geo)
        Y.append(int(y))
        TSTART.append(w["Date"].iloc[0])

X_seq  = np.array(X_seq, dtype=np.float32)
X_geo  = np.array(X_geo, dtype=np.float32)
Y      = np.array(Y, dtype=np.int32)
TSTART = pd.to_datetime(pd.Series(TSTART)).to_numpy()

print("\nBuilt windows:", len(Y), "| seq:", X_seq.shape, "| geo:", X_geo.shape)
print("Label distribution:", pd.Series(Y).value_counts().sort_index().to_dict())

# ---------------------------
# 5) Strict calendar split
# ---------------------------
train_end_time = np.datetime64("2023-01-01T00:00:00")
val_end_time   = np.datetime64("2024-01-01T00:00:00")

order = np.argsort(TSTART)
X_seq, X_geo, Y, TSTART = X_seq[order], X_geo[order], Y[order], TSTART[order]

train_idx = np.where(TSTART < train_end_time)[0]
val_idx   = np.where((TSTART >= train_end_time) & (TSTART < val_end_time))[0]
test_idx  = np.where(TSTART >= val_end_time)[0]

print("\nSplit sizes:", len(train_idx), len(val_idx), len(test_idx))
print("Train time:", pd.to_datetime(TSTART[train_idx[0]]), "->", pd.to_datetime(TSTART[train_idx[-1]]))
print("Val   time:", pd.to_datetime(TSTART[val_idx[0]]),   "->", pd.to_datetime(TSTART[val_idx[-1]]))
print("Test  time:", pd.to_datetime(TSTART[test_idx[0]]),  "->", pd.to_datetime(TSTART[test_idx[-1]]))

Xseq_tr, Xgeo_tr, y_tr = X_seq[train_idx], X_geo[train_idx], Y[train_idx]
Xseq_va, Xgeo_va, y_va = X_seq[val_idx],   X_geo[val_idx],   Y[val_idx]
Xseq_te, Xgeo_te, y_te = X_seq[test_idx],  X_geo[test_idx],  Y[test_idx]

# ---------------------------
# 6) Train-only preprocessing
# ---------------------------
geo_scaler = RobustScaler()
Xgeo_tr_s = geo_scaler.fit_transform(Xgeo_tr)
Xgeo_va_s = geo_scaler.transform(Xgeo_va)
Xgeo_te_s = geo_scaler.transform(Xgeo_te)

# robust-scale only relvol channel globally
REL_CH = 11
rel_scaler = RobustScaler()

rel_tr = Xseq_tr[..., REL_CH].reshape(-1, 1)
rel_va = Xseq_va[..., REL_CH].reshape(-1, 1)
rel_te = Xseq_te[..., REL_CH].reshape(-1, 1)

rel_tr_s = rel_scaler.fit_transform(rel_tr).reshape(Xseq_tr.shape[0], WINDOW)
rel_va_s = rel_scaler.transform(rel_va).reshape(Xseq_va.shape[0], WINDOW)
rel_te_s = rel_scaler.transform(rel_te).reshape(Xseq_te.shape[0], WINDOW)

Xseq_tr_s = Xseq_tr.copy(); Xseq_va_s = Xseq_va.copy(); Xseq_te_s = Xseq_te.copy()
Xseq_tr_s[..., REL_CH] = rel_tr_s
Xseq_va_s[..., REL_CH] = rel_va_s
Xseq_te_s[..., REL_CH] = rel_te_s

y_tr_oh = tf.keras.utils.to_categorical(y_tr, NUM_CLASSES)
y_va_oh = tf.keras.utils.to_categorical(y_va, NUM_CLASSES)
y_te_oh = tf.keras.utils.to_categorical(y_te, NUM_CLASSES)

classes = np.arange(NUM_CLASSES)
cw_vals = compute_class_weight(class_weight="balanced", classes=classes, y=y_tr)
cw_raw = {int(c): float(w) for c, w in zip(classes, cw_vals)}
cw = {k: float(np.sqrt(v)) for k, v in cw_raw.items()}
cw[0] = 1.0
print("\nClass weights raw:", cw_raw)
print("Class weights softened:", cw)

# ---------------------------
# 7) SAME MODEL ARCHITECTURE (CNN + Dilated + 1x Transformer + Geo MLP)
# ---------------------------
def transformer_encoder(x, d_model, num_heads, d_ff, dropout=0.1):
    attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=d_model//num_heads, dropout=dropout)(x, x)
    x = layers.Add()([x, attn])
    x = layers.LayerNormalization()(x)
    ff = layers.Dense(d_ff, activation="relu")(x)
    ff = layers.Dropout(dropout)(ff)
    ff = layers.Dense(d_model)(ff)
    x = layers.Add()([x, ff])
    x = layers.LayerNormalization()(x)
    return x

seq_in = layers.Input(shape=(WINDOW, Xseq_tr_s.shape[-1]), name="seq_in")
geo_in = layers.Input(shape=(Xgeo_tr_s.shape[-1],), name="geo_in")

x = layers.Conv1D(64, 3, padding="same", activation="relu")(seq_in)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.10)(x)

x = layers.Conv1D(64, 5, padding="same", activation="relu")(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.10)(x)

x = layers.Conv1D(64, 3, padding="same", dilation_rate=2, activation="relu")(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.10)(x)

d_model = 64
x = layers.Dense(d_model)(x)
x = transformer_encoder(x, d_model=d_model, num_heads=4, d_ff=128, dropout=0.10)

x = layers.GlobalAveragePooling1D()(x)
x = layers.Dense(64, activation="relu")(x)
x = layers.Dropout(0.10)(x)

g = layers.Dense(64, activation="relu")(geo_in)
g = layers.Dropout(0.20)(g)
g = layers.Dense(32, activation="relu")(g)

z = layers.Concatenate()([x, g])
z = layers.Dense(64, activation="relu")(z)
z = layers.Dropout(0.25)(z)
out = layers.Dense(NUM_CLASSES, activation="softmax")(z)

model = Model([seq_in, geo_in], out)

try:
    opt = tf.keras.optimizers.AdamW(learning_rate=1e-3, weight_decay=1e-4)
except Exception:
    try:
        opt = tf.keras.optimizers.experimental.AdamW(learning_rate=1e-3, weight_decay=1e-4)
    except Exception:
        opt = tf.keras.optimizers.Adam(learning_rate=1e-3)

loss_fn = tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.03)
model.compile(optimizer=opt, loss=loss_fn, metrics=["accuracy"])
model.summary()

# ---------------------------
# 8) Macro-F1 callback + EarlyStopping/ReduceLR
# ---------------------------
def macro_f1_np(y_true_oh, y_pred_prob, num_classes=3):
    y_true = np.argmax(y_true_oh, axis=1)
    y_pred = np.argmax(y_pred_prob, axis=1)
    f1s=[]
    for c in range(num_classes):
        tp = np.sum((y_true==c) & (y_pred==c))
        fp = np.sum((y_true!=c) & (y_pred==c))
        fn = np.sum((y_true==c) & (y_pred!=c))
        prec = tp/(tp+fp+1e-9)
        rec  = tp/(tp+fn+1e-9)
        f1 = 2*prec*rec/(prec+rec+1e-9)
        f1s.append(f1)
    return float(np.mean(f1s))

class MacroF1Callback(tf.keras.callbacks.Callback):
    def __init__(self, val_data):
        super().__init__()
        self.val_data = val_data
        self.best = -1.0
        self.best_weights = None
    def on_epoch_end(self, epoch, logs=None):
        (xs, xg), yv = self.val_data
        pred = self.model.predict([xs, xg], verbose=0)
        f1 = macro_f1_np(yv, pred, NUM_CLASSES)
        logs = logs or {}
        logs["val_macro_f1"] = f1
        print(f" — val_macro_f1: {f1:.4f}")
        if f1 > self.best:
            self.best = f1
            self.best_weights = self.model.get_weights()

mf1_cb = MacroF1Callback(val_data=([Xseq_va_s, Xgeo_va_s], y_va_oh))

early = tf.keras.callbacks.EarlyStopping(
    monitor="val_macro_f1", mode="max", patience=10, restore_best_weights=False
)
reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_macro_f1", mode="max", factor=0.5, patience=4, min_lr=1e-5, verbose=1
)

# ---------------------------
# 9) TRAIN
# ---------------------------
hist = model.fit(
    [Xseq_tr_s, Xgeo_tr_s], y_tr_oh,
    validation_data=([Xseq_va_s, Xgeo_va_s], y_va_oh),
    epochs=60,
    batch_size=64,
    class_weight=cw,
    callbacks=[mf1_cb, early, reduce_lr],
    verbose=1
)

if mf1_cb.best_weights is not None:
    model.set_weights(mf1_cb.best_weights)
print("\nBest val macro-F1:", mf1_cb.best)

# ---------------------------
# 10) EVAL + Confusion + ROC
# ---------------------------
proba = model.predict([Xseq_te_s, Xgeo_te_s], verbose=0)
y_pred = np.argmax(proba, axis=1)

print("\nClassification report:")
print(classification_report(y_te, y_pred, digits=4, target_names=CLASS_NAMES, zero_division=0))

cm = confusion_matrix(y_te, y_pred, labels=[0,1,2])
print("Confusion matrix:\n", cm)

# ---- Confusion matrix (normalized background + raw counts)
cm_norm = cm / (cm.sum(axis=1, keepdims=True) + 1e-12)

fig, ax = plt.subplots(figsize=(5.8, 4.8), dpi=200)
im = ax.imshow(cm_norm, interpolation="nearest")
fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

ticks = np.arange(NUM_CLASSES)
ax.set_xticks(ticks); ax.set_yticks(ticks)
ax.set_xticklabels(CLASS_NAMES, rotation=18)
ax.set_yticklabels(CLASS_NAMES)
ax.set_xlabel("Predicted label")
ax.set_ylabel("True label")

for i in range(NUM_CLASSES):
    for j in range(NUM_CLASSES):
        ax.text(j, i, f"{cm[i,j]}", ha="center", va="center", fontsize=10)

fig.tight_layout()
plt.show()
plt.close(fig)

# ---- ROC (OVR) + AUC
y_te_bin = label_binarize(y_te, classes=[0,1,2])

try:
    auc_ovr = float(roc_auc_score(y_te_bin, proba, multi_class="ovr"))
    print("OVR ROC-AUC:", auc_ovr)
except Exception as e:
    auc_ovr = np.nan
    print("AUC not computed:", e)

fig, ax = plt.subplots(figsize=(6.8, 4.8), dpi=200)
for k, name in enumerate(CLASS_NAMES):
    fpr, tpr, _ = roc_curve(y_te_bin[:, k], proba[:, k])
    ax.plot(fpr, tpr, linewidth=1.6, label=f"{name} (AUC={auc(fpr, tpr):.3f})")

ax.plot([0,1],[0,1], linestyle="--", linewidth=1.0, label="Chance")
ax.set_xlabel("False Positive Rate")
ax.set_ylabel("True Positive Rate")
ax.grid(alpha=0.25)

# legend outside right
handles, labels = ax.get_legend_handles_labels()
fig.subplots_adjust(right=0.80)
ax.legend(handles, labels, loc="center left", bbox_to_anchor=(1.02, 0.5),
          frameon=True, fontsize=8, borderpad=0.25,
          labelspacing=0.2, handlelength=1.1, handletextpad=0.4)

fig.tight_layout()
plt.show()
plt.close(fig)

print("\nDone ✅")
```
